```python
import pandas as pd
data = [
['Y','Y','Y','Y','Positive'],
['Y','N','Y','Y','Positive'],
['N','Y','N','N','Negative'],
['Y','Y','Y','N','Positive'],
['N','N','N','Y','Negative']
]

df = pd.DataFrame(data, columns=['Fever','Cough','Throat_Pain','Body_Pain','Covid19'])

concepts = df.iloc[:, :-1].values
target = df.iloc[:, -1].values

hypothesis = ['0'] * len(concepts[0])

for i in range(len(target)):
    if target[i] == 'Positive':
        if hypothesis[0] == '0':
            hypothesis = concepts[i].copy()
        else:
            for j in range(len(hypothesis)):
                if hypothesis[j] != concepts[i][j]:
                    hypothesis[j] = '?'

print("Final Hypothesis:", hypothesis)
```

```python
import pandas as pd
data = [
['N','Y','Y','Y','Positive'],
['Y','Y','Y','Y','Positive'],
['Y','N','Y','Y','Positive'],
['N','Y','N','N','Negative'],
['Y','Y','Y','N','Positive'],
['N','N','N','Y','Negative'],
['N','N','N','N','Negative']
]

df = pd.DataFrame(data, columns=['Fever','Cough','Throat_Pain','Body_Pain','Covid19'])

concepts = df.iloc[:, :-1].values
target = df.iloc[:, -1].values

S = ['0'] * len(concepts[0])
G = ['?'] * len(concepts[0])

for i in range(len(target)):
    if target[i] == 'Positive':
        if S[0] == '0':
            S = concepts[i].copy()
        else:
            for j in range(len(S)):
                if S[j] != concepts[i][j]:
                    S[j] = '?'

print("Specific Boundary (S):", S)
print("General Boundary (G):", G)
```
