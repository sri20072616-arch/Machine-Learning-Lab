```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score
def get_user_input():
    try:
        attendance = float(input("Enter Attendance (50–100): "))
        internal_marks = float(input("Enter Internal Marks (20–50): "))
        activity_marks = float(input("Enter Activity Marks (5–20): "))

        end_sem_marks = (
            0.4 * attendance +
            0.3 * internal_marks +
            0.3 * activity_marks
        )

        print(f"\n Calculated End Semester Marks: {end_sem_marks:.2f}")
    except ValueError:
        print(" Invalid input. Please enter numeric values.")

get_user_input()


data = pd.DataFrame({
    'Attendance': Attendance,
    'Internal_Marks': internal_marks,
    'Activity_Marks': activity_marks,
    'End_Sem_Marks': end_sem_marks
})

X = data[['Attendance', 'Internal_Marks', 'Activity_Marks']]
y = data['End_Sem_Marks']

model = LinearRegression()
model.fit(X, y)

y_pred = model.predict(X)
r2 = r2_score(y, y_pred)
print(f"R-squared: {r2:.2f}")

plt.figure(figsize=(8, 6))
sns.scatterplot(x=y, y=y_pred)
plt.xlabel("Actual End Semester Marks")
plt.ylabel("Predicted Marks")
plt.title("Actual vs Predicted End Semester Marks")
plt.plot([y.min(), y.max()], [y.min(), y.max()], 'r--')
plt.tight_layout()
plt.show()

def predict_performance():
    try:
        attendance = float(input("Enter Attendance (50–100): "))
        internal_marks = float(input("Enter Internal Marks (20–50): "))
        activity_marks = float(input("Enter Activity Marks (5–20): "))
        
        input_data = pd.DataFrame([[attendance, internal_marks, activity_marks]],
                                  columns=['Attendance', 'Internal_Marks', 'Activity_Marks'])
        predicted_marks = model.predict(input_data)[0]
        print(f"\n Predicted End Semester Marks: {predicted_marks:.2f}")
    except ValueError:
        print(" Invalid input. Please enter numeric values.")

predict_performance()


```
