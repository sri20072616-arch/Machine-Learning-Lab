```python
# ID3 Decision Tree Classifier

import pandas
from sklearn.tree import DecisionTreeClassifier
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split
from sklearn import metrics
from sklearn.metrics import classification_report, confusion_matrix

# 1. Create dataset
data = {
    'CGPA': ['g9','g8','g9','l8','g8','g9','l8','g9','g8','g8'],
    'Inter': ['Y','N','N','N','Y','Y','Y','N','Y','Y'],
    'PK': ['+++','+','==','==','+','+','+','+++','+','=='],
    'CS': ['G','M','P','G','M','M','P','G','G','G'],
    'Job': ['Y','Y','N','N','Y','Y','N','Y','Y','Y']
}

# 2. Create DataFrame
table = pandas.DataFrame(
    data,
    columns=['CGPA', 'Inter', 'PK', 'CS', 'Job']
)

print("Original Dataset:")
print(table)

# 3. Encode categorical values into numbers
encoder = LabelEncoder()

for i in table:
    table[i] = encoder.fit_transform(table[i])

print("\nEncoded Dataset:")
print(table)

# 4. Select input features
X = table.iloc[:, 0:4].values

# 5. Select target variable
y = table.iloc[:, 4].values

# 6. Split dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=2
)

# 7. Create Decision Tree using ID3
model = DecisionTreeClassifier(
    criterion='entropy',
    max_depth=3
)

# 8. Train the model
model.fit(X_train, y_train)

# 9. Predict test data
y_pred = model.predict(X_test)

# 10. Calculate accuracy
print("\nAccuracy:",
      metrics.accuracy_score(y_test, y_pred))

# 11. Display confusion matrix
print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

# 12. Display classification report
print("\nClassification Report:")
print(classification_report(y_test, y_pred))

# 13. Predict new sample
print("\nNew Sample 1:")
print([1, 0, 0, 1])

if model.predict([[1, 0, 0, 1]])[0] == 1:
    print("Got JOB")
else:
    print("Didn't get JOB")

# 14. Predict another new sample
print("\nNew Sample 2:")
print([2, 0, 2, 0])

if model.predict([[2, 0, 2, 0]])[0] == 1:
    print("Got JOB")
else:
    print("Didn't get JOB")
```
